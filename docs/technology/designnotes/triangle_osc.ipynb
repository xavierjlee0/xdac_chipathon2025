```python
# define the given parameters as taken from the specification table or inital guesses
gm_id_p = 5 #current mirror
gm_id_p2 = 9 #p-type switch
gm_id_n = 7.7  #current mirror n

l_p2 = 0.5
l_p = 1
l_n = 1    

# Current Source Limitations
i_bias = 30e-6               #Reference current is 10uA. Note that if some internal design uses something like 30uA, you can mirror that value too
delta_v = 2.5                  #max voltage swing
c_par = 10e-12 #
vdd = 3.3
```

```python
vgs_p = pfet.look_upVGS(GM_ID=gm_id_p, L=l_p, VDS=VDS1, VSB=0.0)
vgs_n = nfet.look_upVGS(GM_ID=gm_id_n, L=l_n, VDS=VDS1, VSB=0.0)
print('vgs_p =', round(float(vgs_p), 3), 'V')
print('vgs_n =', round(float(vgs_n), 3), 'V')

# calculate all widths
print("Current mirrors:")
id_w_p = pfet.lookup('ID_W', GM_ID=gm_id_p, L=l_p, VDS=vgs_p, VSB=0)
w_p = (i_bias) / id_w_p
w_p_round = max(round(w_p*2)/2, 0.5)
print('M_P W =', round(w_p, 2), 'um, rounded W =', w_p_round, 'um')

id_n = (i_bias) 
id_w_n = nfet.lookup('ID_W', GM_ID=gm_id_n, L=l_n, VDS=vgs_n, VSB=0)
w_n = id_n / id_w_n
w_n_round = max(round(w_n*2)/2, 0.5)
print('M_N W =', round(w_n, 2), 'um, rounded W =', w_n_round, 'um')

```

```python
v_min = 2*(2/gm_id_n)
v_max = vdd-(2*(2/gm_id_p))
print('v_min', round(v_min, 3))
print('v_max', round(v_max, 3))


f_max = i_bias/(2*c_par*(v_max-v_min))
print('f_max', round(f_max/1000), 'kHz, actual:', round(f_max), 'Hz')
```
