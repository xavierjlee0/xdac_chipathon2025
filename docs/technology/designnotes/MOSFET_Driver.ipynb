```python
#imported libraries
import numpy as np
import scipy.constants as sc
import matplotlib.pyplot as plt
from pygmid import Lookup as lk
```

```python
#Output Stage Considerations
# 1. Efficiency 
# 2. Voltage Swing
# 3. Output Power
# 4. Power Dissipation
```

```python
#For the 32 ohm optimized system

#Efficiency
#nu = r_l / (r_op + r_on +r_l)
#for now assume r_op = r_on = r_o -- nu = r_l / (2 * r_o +r_l)
r_l = 32
nu = .99 #large target efficiency to account for errors
r_o = (r_l / nu - r_l) / 2
print('on-resistance =', round(r_o, 3), 'ohms')

#Voltage swing - min = 1.4 Vpp
#Using R model to determine the output voltage of the driver
V_min = 1.4
Vdd = 3.3
Vout = Vdd * r_l / (r_l + r_o)
if Vout >= V_min:
 print('Swing is', round(Vout, 3), 'Vpp')
 print('Success')
else:
 print('Swing is', round(Vout, 3), 'Vpp') 
 print('Failure')

#Output Power - target 40mW
Pout = np.square(Vout / (np.sqrt(2) * 2)) / r_l
print('Output Power =', round(Pout, 3) * 1e3, 'mW')

```

```python
#For the 32 ohm optimized system

#Ron = 1 / (uCox * (w / l) * (Vgs - Vth)))
#Nmos width sizing
#Nmos Parameters from nmos_3p3.8 model
l_mos = 0.28 #maximize w/l for given parameters using minimum l
u_n = 0.046929
e_0 = 8.854e-12
e_s = 3.9
tox_n = 8e-9
Cox_n = e_0 * e_s / tox_n
uCox_n = u_n * Cox_n
vth_n = 0.6783
vgs_n = 3.3
Vov_n = vgs_n - vth_n
#width
w_nmos = 1 / (r_o * uCox_n * Vov_n / l_mos)
mosfet_width = 5
print('nmos width =', round(w_nmos, 3), 'um')
print('number of nmos =', round(w_nmos / mosfet_width, 2))

#Pmos width sizing
#Pmos Parameters from pmos_3p3.8 model
u_p = 0.011353
e_0 = 8.854e-12
e_s = 3.9
tox_p = 7.9e-9
Cox_p = e_0 * e_s / tox_p
uCox_p = u_p * Cox_p
vth_p = -0.76227
vgs_p = 3.3
Vov_p = np.abs(-vgs_p - vth_p)
#width
w_pmos = 1 / (r_o * uCox_p * Vov_p / l_mos)
print('pmos width =', round(w_pmos, 3), 'um')
print('number of pmos =', round(w_pmos / mosfet_width, 2))
```
